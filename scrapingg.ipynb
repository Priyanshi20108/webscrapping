### Web Scrapping
Web scraping is the automated process of using software to extract data from websites. Instead of manually copying and pasting information, scrapers fetch web pages and collect specific data points automatically.

```python
# importing required libraries 
from bs4 import BeautifulSoup as bs 
import requests 
import csv 
import lxml
```

```python
# Defining Target URL and HTTP Headers
url = "https://www.amazon.in/Samsung-inches-Smart-LED-UA32H4550FUXXL/dp/B0F84FBWQM/ref=sr_1_4?crid=2LQUFN3I937BX&dib=eyJ2IjoiMSJ9.19o60aazkl1yUhm6cb2CssG8SdL87wYrCJ6Jx_VYF_gCyi6qAKEoI6jbMCw7TWsGzvgGjSkgfwxHnT4iFAwbkhm8kPpfwpHtqUU0bghV5gY5DTOS_vk7YFC65NO0XhvARmfxLreXFg0HDFL2mhm85k-EfDmBmcZDVjR7mlCfmrdobEUiJR6hcvZrD9iL-uauSJeZ3BGFFQV2qznTxiyYDsxW2euB0QUNVNt-REdhK_I.XK8ymPU0qcqPgDlIgqNWFdarlXm46SvU2M9B5UrE2jk&dib_tag=se&keywords=tv&qid=1788430333&sprefix=tv%2Caps%2C250&sr=8-4&th=1"
header = {"user-Agent" : "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/152.0.0.0 Safari/537.36 Edg/152.0.0.0"}
```

```python
##Sending the HTTP GET Request
response = requests.get(url , headers = header )
```

```python
# Validating the Response Statu
if response.status_code == 200 :
    html_content = response.text
else :
    print("error")    
```

```python
#print(html_content) 
```

```python
# Parsing the HTML Content
soup = bs(html_content, "lxml")
```

```python
# Extracting Product Attributes
product_name = soup.find("span",id = "productTitle").text.strip()
product_price = soup.find("span" , class_ = "a-price-whole").text.strip()
product_description = soup.find(id = "feature-bullets").text.strip()
product_size = soup.find("span" , class_ = "a-size-base po-break-word").text.strip() 
product_warranty = soup.find("div" , id = "AMAZON_PROTECT").text.strip()
product_review = soup.find("span" , class_ = "aui-primitive __SAR2l0zNyyuZ").text.strip()
product_features = soup.find("tr" , class_="a-spacing-small po-special_feature").text.strip()
```

```python
# Exporting Data to CSV
with open("details.csv" , mode = "w" , newline = "" , encoding = "utf-8") as file :
    writer = csv.writer(file)
    writer.writerow(["product_name","product_price","product_description","product_size","product_warranty","product_review","product_features"])
    writer.writerow([product_name,product_price,product_description,product_size,product_warranty,product_review,product_features])
```
